```python
import pandas as pd
import numpy as np
```

```python
# Load datasets

nav_df = pd.read_csv("../dataset/raw/02_nav_history.csv")

transactions_df = pd.read_csv("../dataset/raw/08_investor_transactions.csv")

performance_df = pd.read_csv("../dataset/raw/07_scheme_performance.csv")
```

```python
# Check missing values
nav_df.isnull().sum()
```

```python
# Convert date column to datetime
nav_df["date"] = pd.to_datetime(nav_df["date"])
```

```python
# Sort by AMFI code and Date
nav_df = (
    nav_df
    .sort_values(["amfi_code", "date"])
    .reset_index(drop=True)
)
```

```python
# Forward fill missing NAV
nav_df["nav"] = (
    nav_df
    .groupby("amfi_code")["nav"]
    .ffill()
)
```

```python
before = len(nav_df)

nav_df = nav_df.drop_duplicates()

after = len(nav_df)

print(f"Duplicates Removed: {before-after}")
```

```python
invalid_nav = nav_df[nav_df["nav"] <= 0]

print(f"Invalid NAV Records: {len(invalid_nav)}")

invalid_nav.head()
```

```python
nav_df.isnull().sum()
```

```python
nav_df.to_csv(
    "../dataset/processed/nav_history_cleaned.csv",
    index=False
)

print("nav_history_cleaned.csv saved successfully.")
```
