```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
nav_df = pd.read_csv("../dataset/processed/nav_history_cleaned.csv")
transactions_df = pd.read_csv("../dataset/processed/investor_transactions_cleaned.csv")
performance_df = pd.read_csv("../dataset/processed/scheme_performance_cleaned.csv")
```

```python
engine = create_engine("sqlite:///../database/bluestock_mf.db")

print("Database Created Successfully!")
```

```python
nav_df.to_sql(
    "fact_nav",
    engine,
    if_exists="replace",
    index=False
)

transactions_df.to_sql(
    "fact_transactions",
    engine,
    if_exists="replace",
    index=False
)

performance_df.to_sql(
    "fact_performance",
    engine,
    if_exists="replace",
    index=False
)

print("All tables loaded successfully!")
```
