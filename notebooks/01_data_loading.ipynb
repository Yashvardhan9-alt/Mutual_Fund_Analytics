```python
import pandas as pd
from pathlib import Path

# Path to raw dataset
data_path = Path("../dataset/raw")

# Dictionary to store DataFrames
datasets = {}

# Load all CSV files
for file in data_path.glob("*.csv"):
    datasets[file.stem] = pd.read_csv(file)

print(f"✅ Successfully loaded {len(datasets)} datasets.\n")

# Print dataset names
for name in datasets.keys():
    print(name)
```

```python
# Display basic information about each dataset

for name, df in datasets.items():
    print("=" * 60)
    print(f"Dataset : {name}")
    print(f"Shape   : {df.shape}")
    print(f"Columns : {list(df.columns)}")
    print(df.head(3))
    print()
```

```python
# Explore Fund Master Dataset

fund_master = datasets["01_fund_master"]

print("Shape:", fund_master.shape)

print("\nColumns:")
print(fund_master.columns.tolist())

print("\nData Types:")
print(fund_master.dtypes)

print("\nFirst 5 Rows:")
display(fund_master.head())

print("\nMissing Values:")
print(fund_master.isnull().sum())
```
